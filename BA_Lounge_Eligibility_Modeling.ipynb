```python
import pandas as pd
```

```python
df = pd.read_excel('British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx')
```

```python
df.shape
```

```python
df.head()
```

```python
df['TOTAL_SEATS'] = (df["FIRST_CLASS_SEATS"] + df["BUSINESS_CLASS_SEATS"] + df["ECONOMY_SEATS"])
df.head()
```

```python
df["TIER1_PCT"] = (df["TIER1_ELIGIBLE_PAX"] / df["TOTAL_SEATS"]) * 100
df["TIER2_PCT"] = (df["TIER2_ELIGIBLE_PAX"] / df["TOTAL_SEATS"]) * 100
df["TIER3_PCT"] = (df["TIER3_ELIGIBLE_PAX"] / df["TOTAL_SEATS"]) * 100
```

```python
df[
    [
        "FLIGHT_NO",
        "HAUL",
        "TOTAL_SEATS",
        "TIER1_ELIGIBLE_PAX",
        "TIER1_PCT",
        "TIER3_PCT",
    ]
].head()
```

```python
lookup_table = (
    df.groupby(['HAUL','TIME_OF_DAY'])[
    ["TIER1_PCT", "TIER2_PCT", "TIER3_PCT"]
    ]
    .mean()
    .reset_index()
)
```

```python
lookup_table["TOTAL_ELIGIBLE_PCT"] = (
    lookup_table["TIER1_PCT"]
    + lookup_table["TIER2_PCT"]
    + lookup_table["TIER3_PCT"]
)
```

```python
lookup_table["Category"] = (
    lookup_table["HAUL"].replace(
        {"SHORT": "Short-haul", "LONG": "Long-haul"}
    )
    + " "
    + lookup_table["TIME_OF_DAY"]
)
```

```python
numeric_cols = [
    "TIER1_PCT",
    "TIER2_PCT",
    "TIER3_PCT",
    "TOTAL_ELIGIBLE_PCT",
]
lookup_table[numeric_cols] = lookup_table[numeric_cols].round(2)
```

```python
print("============  BRITISH AIRWAYS LOUNGE ELIGIBILITY LOOKUP TABLE ===========")
display(
    lookup_table[
        [
            "Category",
            "TIER1_PCT",
            "TIER2_PCT",
            "TIER3_PCT",
            "TOTAL_ELIGIBLE_PCT",
        ]
    ]
)
```

```python
sample_indices = [1, 9, 8, 2, 0, 7, 3, 4, 5, 6]
sample_df = df.iloc[sample_indices].copy()

sample_df["Category"] = (
    sample_df["HAUL"].replace({"SHORT": "Short-haul", "LONG": "Long-haul"})
    + " "
    + sample_df["TIME_OF_DAY"]
)
```

```python
sample_df["Tier 1 %"] = sample_df["TIER1_PCT"].round(1).astype(str) + "%"
sample_df["Tier 2 %"] = sample_df["TIER2_PCT"].round(1).astype(str) + "%"
sample_df["Tier 3 %"] = sample_df["TIER3_PCT"].round(1).astype(str) + "%"
```

```python
columns_to_show = [
    "FLIGHT_NO",
    "FLIGHT_TIME",
    "ARRIVAL_STATION_CD",
    "Category",
    "Tier 1 %",
    "Tier 2 %",
    "Tier 3 %",
]
print("================== SAMPLE FLIGHT SCHEDULE WITH ESTIMATED ELIGIBILITY =====================")
display(sample_df[columns_to_show])
```
